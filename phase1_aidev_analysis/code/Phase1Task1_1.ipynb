```python
# Label PR type
pr_df["is_agentic"] = True
human_pr_df["is_agentic"] = False

# Keep only relevant columns
agent_cols = ["id", "title", "body", "merged_at", "is_agentic"]
human_cols = ["id", "title", "body", "merged_at", "is_agentic"]

combined_pr_df = pd.concat(
    [
        pr_df[agent_cols],
        human_pr_df[human_cols]
    ],
    ignore_index=True
)

print("Total PRs:", len(combined_pr_df))
```

```python
combined_pr_df["title"] = combined_pr_df["title"].fillna("")
combined_pr_df["body"] = combined_pr_df["body"].fillna("")

combined_pr_df["text"] = (
    combined_pr_df["title"] + " " + combined_pr_df["body"]
).str.lower()

```

```python
quality_keywords = {

    "Security": ["security", "secure", "protection", "verification","Encryption", "credential protection", "Authorization", "access control", "Confidential data"],
    "User Assistance": ["user assistance", "help system", "documentation", "guide"],
    "Performance Efficiency": ["performance efficiency", "performance", "optimization", "optimize"],
    "Compatibility": ["compatibility", "compatible", "interoperability", "integration support", "interface compatibility", "system fit", "cross-system operation"],
    "Maintainability": ["maintainability", "maintainable"],
    "Safety": ["safety", "safe"],
    "Risk Identification": ["risk identification", "risk analysis", "threat detection"],
    "Reliability": ["reliability", "reliable"],
    "Interaction Capability": ["interaction capability", "interaction", "user interaction", "user friendliness"],
    "Operational Constraint": ["operational constraint", "constraint", "limitation"],
    "Integrity": ["integrity", "data integrity", "tamper resistance"],
    "Functional Correctness": ["functional correctness", "correctness", "accuracy", "valid results"],
    "Flexibility": ["flexibility", "flexible", "versatility", "adjustability"],
    "User Error Protection": ["user error protection", "error prevention"],
    "Co-existence": ["co-existence", "coexistence", "shared environment support", "non-interference", "side-by-side operation", "compatibility in shared use"],
    "Availability": ["availability", "uptime", "accessibility", "service continuity", "readiness", "always-on"],
    "Functional Completeness": ["functional completeness", "complete functionality"],
    "Reusability": ["reusability", "reusable"],
    "Modularity": ["modularity", "modular"],
    "Appropriateness Recognizability": ["appropriateness recognizability"],
    "Testability": ["testability", "test coverage"],
    "Fault Tolerance": ["fault tolerance", "fault tolerant", "failure handling", "error survival"],
    "Capacity": ["capacity", "load limit", "throughput capacity", "scale limit", "volume handling", "workload support"],
    "Scalability": ["scalability", "scalable"],
    "User Engagement": ["user engagement", "engagement"],
    "Resource Utilization": ["resource utilization", "cpu usage", "memory usage"],
    "Resistance": ["resistance", "resilient"],
    "Adaptability": ["adaptability", "adaptable", "portability", "configurability", "platform adaptation", "deployment flexibility"],
    "Interoperability": ["interoperability", "integration"],
    "Authenticity": ["authenticity", "authentication", "genuineness", "identity assurance", "verified identity", "legitimacy", "trusted source"],
    "Self-descriptiveness": ["self-descriptiveness", "self descriptive"],
    "Inclusivity": ["inclusivity", "inclusive"],
    "Hazard Warning": ["hazard warning", "warning system", "danger alert"]
}


import re

quality_patterns = {}

for quality, keywords in quality_keywords.items():
    pattern = r"\b(" + "|".join(re.escape(k) for k in keywords) + r")\b"
    quality_patterns[quality] = re.compile(pattern, re.IGNORECASE)
```

```python
def detect_quality(text):
    detected = []
    for quality, keywords in quality_keywords.items():
        if any(keyword in text for keyword in keywords):
            detected.append(quality)
    return detected

combined_pr_df["quality_attributes"] = (
    combined_pr_df["text"].apply(detect_quality)
)
```

```python
def detect_quality(text):
    detected = []

    if not isinstance(text, str):
        return detected

    for quality, pattern in quality_patterns.items():
        if pattern.search(text):
            detected.append(quality)

    return list(set(detected))
```

```python
combined_pr_df["quality_attributes"] = (
    combined_pr_df["text"].apply(detect_quality)
)
```

```python
rq1_df = combined_pr_df.explode("quality_attributes")

# Count PRs per type
total_agentic = combined_pr_df[combined_pr_df["is_agentic"] == True]["id"].nunique()
total_human = combined_pr_df[combined_pr_df["is_agentic"] == False]["id"].nunique()

# Count mentions
freq = (
    rq1_df
    .groupby(["is_agentic", "quality_attributes"])["id"]
    .nunique()
    .unstack(fill_value=0)
)

# Convert to percentage
freq_percent = freq.copy()
freq_percent.loc[True] = (freq.loc[True] / total_agentic) * 100
freq_percent.loc[False] = (freq.loc[False] / total_human) * 100
```

```python
terms = freq_percent.columns

# Sort by total frequency for cleaner plot
sorted_terms = (
    freq_percent.loc[True] + freq_percent.loc[False]
).sort_values(ascending=False).index

freq_percent = freq_percent[sorted_terms]
freq = freq[sorted_terms]
```

```python
import matplotlib.pyplot as plt
import numpy as np

terms = freq_percent.columns
y = np.arange(len(terms))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 10), sharey=True)

# LEFT: Agentic
ax1.barh(y, freq_percent.loc[True], align='center')
ax1.set_title("Terms in Agentic Pull Requests")
ax1.set_xlabel("Percentage (%) of Pull Requests")
ax1.set_yticks(y)
ax1.set_yticklabels(terms)
ax1.invert_xaxis()  # Mirror left
ax1.invert_yaxis()

# Add percentage ticks on top
ax1_top = ax1.twiny()
ax1_top.set_xlim(ax1.get_xlim())
ax1_top.set_xticks(ax1.get_xticks())
ax1_top.set_xticklabels([f"{abs(x):.0f}%" for x in ax1.get_xticks()])

# RIGHT: Human
ax2.barh(y, freq_percent.loc[False], align='center', color='orange')
ax2.set_title("Terms in Human Pull Requests")
ax2.set_xlabel("Percentage (%) of Pull Requests")
ax2.set_yticks(y)
ax2.set_yticklabels([])

# Middle Axis for Terms
ax_middle = ax1.twinx()
ax_middle.set_ylim(ax1.get_ylim())
ax_middle.set_yticks(y)
ax_middle.set_yticklabels(terms)
ax_middle.tick_params(axis='y', length=0)


# Add percentage ticks on top
ax2_top = ax2.twiny()
ax2_top.set_xlim(ax2.get_xlim())
ax2_top.set_xticks(ax2.get_xticks())
ax2_top.set_xticklabels([f"{x:.0f}%" for x in ax2.get_xticks()])

for ax in [ax1, ax2]:
    ax.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.savefig("rq3-4_plot.pdf", format="pdf", bbox_inches='tight')
plt.show()
```
